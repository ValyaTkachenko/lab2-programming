## Крок 1. Завантаження датасету

```python
!head -5 DS6.txt
```

## Крок 2. Зчитування датасету

```python
def read_dataset(filename):
    points = []
    with open(filename) as f:
        for line in f:
            parts = line.split()
            if len(parts) == 2:
                x = int(parts[0])
                y = int(parts[1])
                points.append((x, y))
    return points

points = read_dataset("DS6.txt")
print("Кількість точок:", len(points))
print("Перші 5 точок:", points[:5])
```

## Крок 3. Створення полотна 960×540

```python
from PIL import Image, ImageDraw

WIDTH = 960
HEIGHT = 540

image = Image.new("RGB", (WIDTH, HEIGHT), "white")
draw = ImageDraw.Draw(image)

image
```

## Крок 4. Відображення точок

```python
def to_canvas(x, y):
    return x, HEIGHT - 1 - y

for x, y in points:
    draw.point(to_canvas(x, y), fill="black")

image
```

## Крок 5. Збереження результату

```python
image.save("result.png")
print("Зображення збережено у файл result.png")
```

## Висновок
У ході роботи розроблено програму мовою Python, яка зчитує датасет з файлу DS6.txt (53 542 точки), відображає точки на полотні 960×540 пкс із початком координат у нижньому лівому куті та зберігає результат у файл result.png.

Використана бібліотека **Pillow (PIL)** та методи:
- `Image.new()` — створення полотна 960×540 з білим тлом;
- `ImageDraw.Draw()` — створення об'єкта для малювання;
- `draw.point()` — відображення точки (одного пікселя);
- `image.save()` — збереження зображення у формат PNG.

Для зчитування файлу використано стандартні засоби Python: `open()`, `split()`, `int()`.
